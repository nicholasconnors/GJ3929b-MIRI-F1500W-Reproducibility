```python
import numpy as np
import matplotlib.pyplot as plt
```

```python
chain = np.load("./full_chain.npy")
```

```python
fp = chain[-1000:,:,0].flatten() * 1e6 #ppm
t_sec = chain[len(chain)//2:,:,4].flatten() * 24 * 60 # minutes
```

```python
plt.hist(fp, bins="fd")
plt.title(f"fp = {np.median(fp):.0f} +/- {np.std(fp):.0f}")
plt.axvline(np.median(fp), color='black', linestyle='--')
```

```python
plt.hist(t_sec, bins="fd")
plt.title(f"t_sec = {np.median(t_sec):.0f} +/- {np.std(t_sec):.0f}")
plt.axvline(np.median(t_sec), color='black', linestyle='--')
```
